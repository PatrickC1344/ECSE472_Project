```python
import numpy as np
from pathlib import Path
```

```python
LONG_COMP = ['VCVS', 'CCVS', 'VCCS', 'CCCS']
BASIC = ['R','V','I','C','L']
def get_type(name):
    name = name.upper()

    for prefix in LONG_COMP:
        if name.startswith(prefix):
            return prefix
    
    return name[0]
    
def read_netlist(file):
    comps = []
    dirs = []
    seen_names = []
    found_end = False
    with open(file, 'r') as f:
        title = f.readline()

        for line in f:
            line = line.strip()

            #skip comments
            if not line or line.startswith('*'):
                continue
                
            #check for end of netlist
            if line == '.end':
                found_end = True
                break

            #add directives
            if line.startswith('.'):
                dirs.append(line)
                continue
                

            parts = line.split()
            name = parts[0]

            #check for duplicate components
            if name in seen_names:
                raise ValueError('Duplicate components', name)
            seen_names.append(name)

            #check for unsupported types
            comp_type = get_type(name)
            if comp_type not in BASIC and comp_type not in LONG_COMP:
                raise ValueError('Unsupported component type', comp_type)
                
            comps.append(parts)
            
    if found_end == False:
        raise ValueError('Missing .end')

    if len(dirs) != 1:
        raise ValueError('Expected exactly one active directive')

    if len(comps) <= 0:
        raise ValueError('Cannot calculate circuit with no components')
    return title, comps, dirs
```

```python
def gen_matrices(comps):
    node_index = {}
    cur_index = {}
    size = 0
    
    for comp in comps:
        name = comp[0]
        comp_type = get_type(name)

        nodes = comp[1:3]
        if comp_type in LONG_COMP:
            nodes = nodes + comp[3:5]

        for node in nodes:
            if node != "0" and node not in node_index:
                node_index[node] = size
                size += 1 

        if comp_type == 'V' or comp_type == 'VCVS' or comp_type == 'CCCS' or comp_type == 'L':
            cur_name = 'I'+name
            cur_index[cur_name] = size
            size += 1
        elif comp_type == 'CCVS':
            cur_name = 'I'+name
            cur_index[cur_name] = size
            size += 1
            cur_ctrl = 'Ix'+name
            cur_index[cur_ctrl] = size
            size += 1
        
    G = np.zeros((size, size), dtype=float)
    C = np.zeros((size, size), dtype=float)
    b = np.zeros(size, dtype=float)
    return G, C, b, node_index, cur_index
```

```python
def src_parse(comp):
    if len(comp) < 4:
        raise ValueError('Missing source value', comp[0])

    if comp[3].upper() == 'DC':
        if len(comp) != 5:
            raise ValueError('expected name node+ node- DC value', comp[0])
        value = float(comp[4])
    else:
        if len(comp) != 4:
            raise ValueError('expected name node+ node- value',comp[0])
        value = float(comp[3])

    if not np.isfinite(value):
        raise ValueError('source value must be finite', comp[0])
    return value

    
def fill_matrices(comps, G, C, b, node_index, cur_index):

    def add_G(row, col, value):
        if row is not None and col is not None:
            G[row, col] += value

    def add_C(row, col, value):
        if row is not None and col is not None:
            C[row, col] += value
    
    def add_b(row, value):
        if row is not None:
            b[row] += value
    
    def stamp_voltage_branch(i, j, current):
        add_G(j, current, -1)
        add_G(current, j, -1)
        add_G(i, current, 1)
        add_G(current, i, 1)

    
    G.fill(0.0)
    C.fill(0.0)
    b.fill(0.0)

    for comp in comps:
        name = comp[0]
        comp_type = get_type(name)

        i = node_index[comp[1]]
        j = node_index[comp[2]]

        if comp_type == 'R':
            r = float(comp[3])

            if not np.isfinite(r) or r <= 0:
                raise ValueError('Resistance must be finite and positive')

            g = 1.0 / r

            add_G(i, i, g)
            add_G(i, j, -g)
            add_G(j, i, -g)
            add_G(j, j, g)

        elif comp_type == 'I':
            current = src_parse(comp)

            add_b(i, -current)
            add_b(j, current)

        elif comp_type == 'V':
            voltage = src_parse(comp)
            ie = cur_index['I' + name]

            stamp_voltage_branch(i, j, ie)
            add_b(ie, voltage)

        elif comp_type == 'C':
            F = float(comp[3])

            if not np.isfinite(F) or F < 0:
                raise ValueError('Capacitance must be finite and non-negative')

            add_C(i, i, F)
            add_C(i, j, -F)
            add_C(j, i, -F)
            add_C(j, j, F)

        elif comp_type == 'L':
            H = float(comp[3])
            if not np.isfinite(H) or H < 0:
                raise ValueError('Inductance must be finite and non-negative')
                
            il = cur_index['I' + name]

            stamp_voltage_branch(i, j, il)
            add_C(il, il, -H)

        elif comp_type == 'VCCS':
            cplus = node_index[comp[3]]
            cminus = node_index[comp[4]]
            gm = float(comp[5])

            if not np.isfinite(gm):
                raise ValueError('VCCS current must be finite')

            add_G(i, cplus, gm)
            add_G(i, cminus, -gm)
            add_G(j, cplus, -gm)
            add_G(j, cminus, gm)

        elif comp_type == 'VCVS':
            cplus = node_index[comp[3]]
            cminus = node_index[comp[4]]
            A = float(comp[5])
            ie = cur_index['I' + name]

            if not np.isfinite(A):
                raise ValueError('VCVS current must be finite')

            stamp_voltage_branch(i, j, ie)
            add_G(ie, cplus, -A)
            add_G(ie, cminus, A)

        elif comp_type == 'CCCS':
            cplus = node_index[comp[3]]
            cminus = node_index[comp[4]]
            A = float(comp[5])
            isense = cur_index['I' + name]

            if not np.isfinite(A):
                raise ValueError('CCCS current must be finite')

            stamp_voltage_branch(cplus, cminus, isense)
            add_G(i, isense, A)
            add_G(j, isense, -A)

        elif comp_type == 'CCVS':
            cplus = node_index[comp[3]]
            cminus = node_index[comp[4]]
            A = float(comp[5])

            if not np.isfinite(A):
                raise ValueError('CCVS current must be finite')

            ie = cur_index['I' + name]
            isense = cur_index['Ix' + name]

            stamp_voltage_branch(cplus, cminus, isense)
            stamp_voltage_branch(i, j, ie)

            add_G(ie, isense, -A)

        else:
            raise ValueError(f"Unsupported component: {name}")

    return G, C, b
```

```python
name, comps, dirs = read_netlist('data.txt')
G, C, b, node_index, cur_index = gen_matrices(comps)
node_index['0'] = None
fill_matrices(comps, G, C, b, node_index, cur_index)
print(node_index)
print(cur_index)
print(G)
print(C)
print(b)
```

```python
def lin_dc_analysis():
    try: 
        x = np.linalg.solve(G, b)
    except np.linalg.LinAlgError as e:
        raise ValueError('No unique DC Solution') from e
    for key in node_index:
        if key == '0':
            continue
        index = node_index[key]
        val = x[index]
        print(key + ":" + str(val) + ' V')
    for key in cur_index:
        index = cur_index[key]
        val = x[index]
        print(key + ":" + str(val) + ' A')
        
if dirs[0].lower() == '.dc':
    lin_dc_analysis()
else:
    print('Currently can only do linear dc analysis')
```
